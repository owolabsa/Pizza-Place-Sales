```python
#Joins sd1 and sd2 using outer join

sd5 = pd.merge(sd1, sd2, on='order_id', how='outer')
sd5.head()
```

```python
#Joins sd3 and sd4 using outer join

sd6 = pd.merge(sd3, sd4, on='pizza_type_id', how='outer')
sd6.head()
```

```python
#Joins sd5 and sd6 using outer join

sd7 = pd.merge(sd5, sd6, on='pizza_id', how='outer')
sd7.head()
```

```python
sd7.info()
```

```python
#check for null value

sd7.isna().sum()
```

```python
sd7["quantity"].value_counts()
```

```python
#to handle null value, fill with mean 

sd7["quantity"] = sd7["quantity"].fillna(sd7["quantity"].mean())
sd7["order_id"] = sd7["order_id"].fillna(sd7["order_id"].mean())
sd7["order_details_id"] = sd7["order_details_id"].fillna(sd7["order_details_id"].mean())
```

```python
sd7.isna().sum()
```

```python
#covert transaction date and time to date_time dataset

sd7["date"] = pd.to_datetime(sd7["date"])
sd7["time"] = pd.to_datetime(sd7["time"])
```

```python
sd7["date"].value_counts()
```

```python
#to handle null value

sd7["date"] = sd7["date"].replace(pd.NA)
sd7["time"] = sd7["time"].replace(pd.NA)
```

```python
sd7.isna().sum()
```

```python
sd7.info()
```

```python
#check for duplicate

sd7.duplicated().sum()
```

```python
sd7.head(10)
```

```python
#To calculate sales

sd7["sales"] = sd7["quantity"] * sd7["price"]
```

```python
sd7.head()
```

# Total Revenue/Sales

```python
Total_revenue =sd7["sales"].sum()

Total_revenue
```

# Total Quantity Sold

```python
total_quantity = sd7["quantity"].sum()

total_quantity
```

```python
total_orders = sd7["order_id"].count()

total_orders
```

# Pizzer Type Sold

```python
pizza_type = sd7["pizza_type_id"].nunique()

pizza_type
```

# Average Price of the Pizzas

```python
average_price = sd7["price"].mean()

average_price
```

# Peak Hour of Sales

```python
sales_by_hour = sd7.groupby("time")["sales"].sum().sort_values(ascending=False).idxmax()

sales_by_hour
```

# Sales on each Day of the Week

```python
sd7["day_in_words"] = sd7["time"].dt.day_name()

sd7.head()
```

# Day of The Week When Sales are Made The Most?

```python
sales_by_day = sd7.groupby('day_in_words')['sales'].sum().sort_values(ascending=False)

sales_by_day
```

# Top 5 bestselling pizzas

```python
top_pizzas = sd7.groupby('name')['quantity'].sum().sort_values(ascending=False)
top_pizzas.head(5)
```

# Lowest Selling Pizzas

```python
low_sales_pizzas = sd7.groupby('name')['quantity'].sum().sort_values().head(5)
print(low_sales_pizzas)
```

# Visualisation

# Peak Sales Hour

```python
plt.bar(sd7.order_id, sd7.sales)
plt.title("Sales by Day")
plt.xlabel("sales")
plt.ylabel("order_id")
plt.show()
```

```python
plt.bar(top_pizzas.index, top_pizzas.values)
plt.title("Top 5 Best-Selling Pizzas")
plt.xlabel("Pizza Name")
plt.ylabel("Quantity Sold")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.bar(sales_by_day.index, sales_by_day.values)
plt.title("Sales by Day of the Week")
plt.xlabel("Day")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```
